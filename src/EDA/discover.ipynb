### This notebook is to discover our dataset further and potentially improve

```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('../../data/combined_data.csv', index_col='Unnamed: 0')
```

```python
df.head()
```

```python
shapes = df['shape'].unique()
print('Number of unique shapes: ',len(shapes))
print(shapes)
```

```python
shape_coordinates = df['shape_id'].value_counts()
print('shape with most coordinates: ', shape_coordinates.max())
print('shape with least coordinates: ', shape_coordinates.min())
```

```python
# Label distribution
label_counts = df['label'].value_counts()
print("Label Distribution:")
print(f"  Label 1 (Line): {label_counts.get(1, 0)} points")
print(f"  Label 2 (Arc):  {label_counts.get(2, 0)} points")
print(f"\nPercentage:")
print(f"  Label 1: {label_counts.get(1, 0) / len(df) * 100:.2f}%")
print(f"  Label 2: {label_counts.get(2, 0) / len(df) * 100:.2f}%")
label_counts
```
